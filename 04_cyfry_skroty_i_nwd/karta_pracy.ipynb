## Pozycja cyfry i podstawa systemu

W systemie o podstawie p cyfry mają wartości od 0 do p-1. Wagi od prawej to 1,p,p²,... . Zapis 1011₂ oznacza 1·8+0·4+1·2+1=11. 102₃ oznacza 1·9+0·3+2=11. W systemie szesnastkowym A–F oznaczają 10–15.

Do wydzielenia ostatniej cyfry dowolnego systemu służy n % p; n // p usuwa tę cyfrę. Zero wymaga osobnej uwagi: ma jedną cyfrę, chociaż while n > 0 nie wykona się ani razu.


```python
n = 45
reszty = []
while n > 0:
    reszty.append(n % 2)
    n //= 2
print(reszty)  # cyfry od najmniej znaczącej
print(list(reversed(reszty)))
```

## Horner i budowanie liczby

## Algorytm Euklidesa

Każdy wspólny dzielnik a i b dzieli również resztę a % b. Zastąpienie pary (a,b) przez (b,a % b) zachowuje NWD, a druga liczba maleje, aż stanie się zerem.

Na końcu a jest wynikiem. Dla dowolnych znaków argumentów najpierw bierzemy wartości bezwzględne. Dla b=0 wynikiem jest |a|. W kursie NWD(0,0) przyjmujemy jako 0. W 3.1 matury 2024 ograniczenia dotyczą operacji wewnątrz funkcji skrótu: stosujemy tam wyłącznie arytmetykę całkowitą.


```python
a, b = 84, 35
while b != 0:
    print(a, b, a % b)
    a, b = b, a % b
print("NWD:", a)
```

## Zadanie 13: NWW

Napisz funkcję `nww(a, b)`, która dla dwóch nieujemnych liczb całkowitych zwróci ich najmniejszą wspólną wielokrotność. Dla dodatnich argumentów jest to **najmniejsza dodatnia liczba, która dzieli się bez reszty zarówno przez `a`, jak i przez `b`**.

Funkcja ma zwracać jedną liczbę całkowitą, nie listę wielokrotności. Jeśli przynajmniej jeden argument jest zerem, przyjmij wynik `0`, również dla `(0, 0)`. W rozwiązaniu wykorzystaj NWD. Wszystkie obliczenia wykonaj na liczbach całkowitych, bez dzielenia zmiennoprzecinkowego.

**Przykład:** dodatnie wielokrotności 4 to 4, 8, 12, 16, …, a wielokrotności 6 to 6, 12, 18, … . Najmniejszą wspólną liczbą jest 12, dlatego `nww(4, 6)` ma zwrócić `12`. To inne pojęcie niż NWD: największy wspólny dzielnik 4 i 6 wynosi 2.

Wyświetl wynik dla wybranej pary dodatnich liczb, pary z jednym zerem i pary `(0, 0)`.


## Zadanie 14: zapis szesnastkowy

Napisz funkcję `szesnastkowo(n)`, która dla nieujemnej liczby całkowitej `n` zwróci jej zapis w systemie szesnastkowym. W tym systemie wartości cyfr 0–9 zapisujemy zwykłymi cyframi, a wartości 10–15 literami A–F.

Funkcja ma **zwrócić napis** złożony ze znaków `0123456789ABCDEF`. Używaj wielkich liter. Nie dodawaj początkowego oznaczenia `0x` ani zbędnych zer na początku. Wyjątek stanowi sama liczba 0, której zapisem jest napis `"0"`.

W rozwiązaniu wykorzystaj własną funkcję `cyfry_w_bazie` z zadania 6. Nie używaj `hex` do tworzenia wyniku tej funkcji. Pamiętaj, że zadanie 6 zwracało listę wartości cyfr, a tutaj wymagany jest ich zapis połączony w jeden napis.

**Przykład:** `szesnastkowo(26)` ma zwrócić `"1A"`, ponieważ `1·16 + 10 = 26`. Wynikami zgodnymi z poleceniem nie są `"0x1a"`, `"1a"`, `"001A"` ani lista `[1, 10]`.

Wyświetl wyniki dla `0` i kilku wybranych liczb dodatnich. Poza własną funkcją możesz porównać je z wynikiem `hex`, uwzględniając różnicę w wielkości liter i obecność oznaczenia `0x`.
