## Sortowanie z zachowaniem wystąpień

sorted tworzy nową uporządkowaną listę, a lista.sort zmienia obecną listę i zwraca None. reverse=True oznacza porządek malejący. Pozycja k liczona od 1 ma indeks k-1.

## Istnieje a dla każdego

any(warunki) zwraca True, gdy choć jeden warunek jest prawdziwy; all wymaga wszystkich. Odpowiednikiem any jest pętla zakończona break po znalezieniu pierwszego dopasowania.

W zadaniu o dzielnikach każdy element pierwszego wiersza liczymy najwyżej raz, nawet jeśli dzieli pięć liczb z drugiego wiersza. Jednak dwa wystąpienia tego samego elementu w pierwszym wierszu liczymy oddzielnie.


````python
pierwszy = [2, 2, 3, 7]
drugi = [12, 15]
wynik = sum(any(n % p == 0 for n in drugi) for p in pierwszy)
print(wynik)
````

## Zapas czynników i niezależność prób

Rozkład na czynniki pierwsze mówi, ile razy każda liczba pierwsza jest potrzebna. Counter(lista) buduje słownik dostępnych liczności. Można też przejść po całej liście czynników i dzielić pozostałą liczbę tylko raz na wystąpienie.

Gdy pozostało 1, liczba jest zbudowana. Gdy nie da się już dzielić i pozostało więcej niż 1, brakuje czynników. Każdą nową liczbę sprawdzamy z pełnym zapasem, bo zadanie nie wymaga jednoczesnego budowania wszystkich liczb.


````python
from collections import Counter
dostepne = Counter([2, 2, 3, 5])
potrzebne = Counter([2, 2, 2, 2])
print(dostepne, potrzebne)
print(potrzebne[2] <= dostepne[2])
````

## Zadanie 12: liczby z dostępnych czynników

Napisz funkcję `mozliwe_liczby(pierwsze, granica)`, która poda **wszystkie liczby całkowite od 2 do `granica` włącznie, możliwe do otrzymania z podanego zapasu czynników pierwszych**. Argument `pierwsze` jest listą liczb pierwszych z możliwymi powtórzeniami, a `granica` jest liczbą całkowitą co najmniej 2.

Obowiązują zasady z zadania 7: liczbę otrzymujemy przez pomnożenie wybranych elementów listy, a każdy element można wykorzystać najwyżej raz. Można też wybrać tylko jeden element. **Każdą liczbę rozpatruj niezależnie**, z całym zapasem od nowa — umieszczenie jednej liczby w odpowiedzi nie zużywa czynników potrzebnych do uzyskania innej.

Funkcja ma **zwrócić listę możliwych liczb uporządkowaną rosnąco, bez powtórzeń**. Nie umieszczaj w niej 1 ani liczb większych od granicy. Jeśli nie da się otrzymać żadnej liczby z podanego zakresu, zwróć `[]`. Możesz korzystać z funkcji `mozna` napisanej wcześniej.

**Przykład:** dla zapasu `[2, 3]` i granicy 8 wynikiem jest `[2, 3, 6]`. Możliwe wybory to sama dwójka, sama trójka lub ich iloczyn. Liczba 4 nie pasuje, bo wymaga dwóch dwójek, a w zapasie jest tylko jedna. Dla tego samego zapasu i granicy 5 wynikiem byłoby `[2, 3]`, ponieważ 6 przekracza granicę.
